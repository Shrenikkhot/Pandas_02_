```python
import pandas as pd
#melt()
#example
wide_df = pd.DataFrame(
    {
        'Store' : ['North','South','East'],
        '2021':[1200,800,900],
        '2022':[1500,900,1100],
        '2023':[1400,850,1000]
    }
)
wide_df
```

```python
#unpivot pd.melt()
long_df = pd.melt(
    wide_df,#data
    id_vars=['Store'], #columns
    value_vars=['2021','2022'], #colum to melt
    var_name='year', #new column
    value_name='sales' #new column

)
long_df
```

```python
long_df.to_csv("data.csv")
```

```python
data = pd.DataFrame(
    {
        'Store' : ['North','South','East'],
        'Product':['Laptop','Mouse','Keyboard'],
        '2021':[1200,800,900],
        '2022':[1500,900,1100],
        '2023':[1400,850,1000],
        'price_day1':[120,450,650],
        'price_day2':[560,890,780]
    }
)
data
```

```python
long_df=pd.melt(
   data,
   id_vars=['Product'],
   value_vars=['price_day1','price_day2'],
   var_name='price day',
   value_name='price'
)
long_df
```

```python
#MELT ALL
long_all = pd.melt(
    wide_df,
    id_vars=['Store'],
    var_name='Year',
    value_name='Sales'
    )
long_all
```

```python
long_all1 = pd.melt(
    data,
    id_vars=['Product'],
    var_name='year',
    value_name='sales'
)
long_all1
```

```python
wide_df['Manager']=['shrenik','rupali','ashok']
long_df = pd.melt(
    wide_df,#data
    id_vars=['Store','Manager'], #columns
    value_vars=['2021','2022','2023'], #colum to melt
    var_name='year', #new column
    value_name='sales' #new column

)
long_df
```

```python
#MELT ALL
long_all = pd.melt(
    data,
    id_vars=['Store','Product'],
    var_name='Year',
    value_name='Sales'
    )
long_all
```

```python
scores_wide = pd.DataFrame({
    'Student': ['Ali', 'Beth', 'Cathy'],
    'Math': [85, 90, 78],
    'Science': [88, 92, 80],
    'English': [90, 85, 88]
})
scores_wide
```

```python
#output
#student | Subject | Score
Student_all=  pd.melt(
    scores_wide,
    id_vars=['Student'],
    var_name='Subject',
    value_name='Score'

)
Student_all
```

```python
#average score per subject 
Student_all.groupby('Subject')['Score'].mean().round(2)
```
